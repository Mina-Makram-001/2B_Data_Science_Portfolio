```python
import os
from google import genai
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Initialize the client (automatically uses OS environment variable)
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

chat = client.chats.create(model="gemini-3.1-flash-lite")
```

```python
prompt = """
give me a one article summary of this essay

# The Essay

Data science has become one of the most important fields in modern technology because organizations increasingly depend on data to make better decisions. A data scientist combines statistics, programming, mathematics, and domain knowledge to transform raw data into useful information. The typical data science process begins with collecting data from different sources and cleaning it by handling missing values, duplicates, inconsistent formats, and outliers.
After cleaning, exploratory data analysis is performed to understand distributions, relationships, and patterns within the dataset. Visualization tools such as Matplotlib, Seaborn, Tableau, and Power BI can help communicate these findings clearly. Data scientists may then engineer new features and transform variables to improve the quality of the analysis.
Machine learning is another important component of data science. Supervised learning algorithms such as linear regression, decision trees, random forests, and gradient boosting can be used for prediction and classification. Unsupervised techniques such as clustering and principal component analysis can help discover hidden structures in data.
Python and SQL are widely used in data science. Python libraries such as Pandas, NumPy, and Scikit-learn provide powerful tools for data manipulation and machine learning, while SQL is essential for working with databases.
Ultimately, successful data science requires more than technical skills. It also requires critical thinking, communication, and the ability to translate analytical results into meaningful real-world insights.
"""

response = client.models.generate_content(
    model="gemini-3.1-flash-lite", contents=prompt
)

print(response.text)
```
